```python
Ufa, Ufb, Ufc = np.array([30.0, 29.0, 29.3], dtype=np.float32)

Rh = 1000
f = 50
w = 2 * np.pi * f
```

```python
def calculate_Uh(Uf, Cf, Ra, Rb, Rc, Rzaz=None):
    a = (Uf / 2)
    b = 9*((1/Rb + 1/Rc)**2) + (np.sqrt(3)*(1/Rb - 1/Rc) + 6*w*Cf)**2
    
    if Rzaz is None:
        c = (1/Rb + 1/Rc + 1/Ra + 1/Rh)**2 + (3*w*Cf)**2
    else:
        c = (1/Rb + 1/Rc + 1/Ra + 1/Rh + 1/Rzaz)**2 + (3*w*Cf)**2

    return a * np.sqrt(b / c)
```

# 3. Построение векторных диаграмм

```python
from tqdm import tqdm
```

```python
def get_U01_calculated_by_phase_voltages(Ufa, Ufb, Ufc, Uf0, Ufa0, Ufb0, Ufc0, delta, epsilon):
    Uf_max = max(Ufa, Ufb, Ufc)

    points_1d = np.arange(-np.floor(Uf_max), np.ceil(Uf_max) + 1e-5, delta)
    points = np.stack(np.meshgrid(points_1d, points_1d), axis=-1).reshape(-1, 2)

    Ufa_point = (0, Ufa)
    Ufb_point = (Ufb * np.cos(2/3), -Ufb * np.sin(2/3))
    Ufc_point = (-Ufc * np.cos(2/3), -Ufc * np.sin(2/3))

    max_length_error = delta / 2

    found_points = []
    for point in tqdm(points, desc="Search..."):
        x = point[0]; y = point[1]

        Uf0_current = np.sqrt(x**2 + y**2)
        Ufa0_current = np.sqrt((x - Ufa_point[0])**2 + (y - Ufa_point[1])**2)
        Ufb0_current = np.sqrt((x - Ufb_point[0])**2 + (y - Ufb_point[1])**2)
        Ufc0_current = np.sqrt((x - Ufc_point[0])**2 + (y - Ufc_point[1])**2)

        Uf0_equal = abs(Uf0 - Uf0_current) <= (max_length_error + epsilon)
        Ufa0_equal = abs(Ufa0 - Ufa0_current) <= (max_length_error + epsilon)
        Ufb0_equal = abs(Ufb0 - Ufb0_current) <= (max_length_error + epsilon)
        Ufc0_equal = abs(Ufc0 - Ufc0_current) <= (max_length_error + epsilon)

        if Uf0_equal and Ufa0_equal and Ufb0_equal and Ufc0_equal:
            found_points.append(point)

    return np.array(found_points)
```

```python
def set_up_vector_diagram_axis(ax, Ufa, Ufb, Ufc):
    ax.set_aspect('equal', adjustable='box')

    Ufa_point = (0, Ufa)
    Ufb_point = (Ufb * np.cos(2/3), -Ufb * np.sin(2/3))
    Ufc_point = (-Ufc * np.cos(2/3), -Ufc * np.sin(2/3))

    for point, letter, hor_align in zip([Ufa_point, Ufb_point, Ufc_point], ['A', 'B', 'C'], ['center', 'left', 'right']):
        ax.annotate(
            '',
            xy = point,
            xytext = (0, 0),
            arrowprops=dict(
                arrowstyle='->',
                color='black',
                lw=2
            )
        )
        ax.text(point[0], point[1], fr"$\dot{{U}}_{{{letter}}}$",
            fontsize=11, ha=hor_align, va='bottom',
            backgroundcolor='white',
            bbox=dict(facecolor='white', alpha=0.0, edgecolor='none', pad=1)
            )
    
    Uf_max = max(Ufa, Ufb, Ufc)
    ax.set_xlim(-Uf_max, Uf_max)
    ax.set_ylim(-Uf_max * 0.8, Uf_max + 5)
```

```python
def set_U01_on_vector_diagram(ax, Ufa, Ufb, Ufc, U01):
    Ufa_point = (0, Ufa)
    Ufb_point = (Ufb * np.cos(2/3), -Ufb * np.sin(2/3))
    Ufc_point = (-Ufc * np.cos(2/3), -Ufc * np.sin(2/3))

    ax.annotate(
            '',
            xy = U01,
            xytext = (0, 0),
            arrowprops=dict(
                arrowstyle='->',
                color='black',
                lw=2,
            )
        )

    for point_to, point_from in zip(
        [Ufa_point, Ufb_point, Ufc_point],
        [U01, U01, U01]
        ):

        ax.annotate(
            '',
            xy = point_to,
            xytext = point_from,
            arrowprops=dict(
                arrowstyle='->',
                color='red',
                lw=1.5,
                linestyle='dashed'
            )
        )
```

```python
# Векторные диаграммы для таблицы №1

# delta отвечает за точность попадания (0.2 вполне достаточно для графика)
# epsilon подбирается, пока массив точек будет непустым
df_idx  = 0 * 5 + 0
delta   = 0.2
epsilon = 1.1

# ================================================================================

Uf0, Ufa0, Ufb0, Ufc0   = data1[df_idx][4:8]
Cf, Rf                  = data1[df_idx][0:2]

found_points = get_U01_calculated_by_phase_voltages(Ufa, Ufb, Ufc, Uf0, Ufa0, Ufb0, Ufc0, delta, epsilon)
print(f"Number of found points: {len(found_points)}")

fig, ax = plt.subplots(1, 2, figsize=(10, 4))

set_up_vector_diagram_axis(ax[0], Ufa, Ufb, Ufc)
set_up_vector_diagram_axis(ax[1], Ufa, Ufb, Ufc)

if len(found_points) > 0:
    ax[0].scatter(found_points[:, 0], found_points[:, 1], color='red', s=5)
    set_U01_on_vector_diagram(ax[1], Ufa, Ufb, Ufc, np.mean(found_points, axis=0))

if df_idx < 5*5:
    ax[1].set_title(fr"$C_ф={Cf:.1f} \hspace{{0.25}} мкФ; R_ф={int(Rf)} \hspace{{0.25}} кОм$")
else:
    ax[1].set_title(f"Закмыкание фазы B на землю:\n" + fr"$ C_ф={Cf:.1f} \hspace{{0.25}} мкФ; R_ф={int(Rf)} \hspace{{0.25}} кОм$")

plt.tight_layout()
plt.show()
```
